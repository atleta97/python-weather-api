```python
import requests

def get_Location():
    
    forbidden = "[]{}<>\/#"

    while True:
        city = input("Input city: ")
        
    
        if city.strip() == "":
            print("City cannot be blank!")

        elif any(char in city for char in forbidden):
            print("City can only contain letters and spaces!")

        else:
            return city

def get_Coords():
    while True:
        location = get_Location()
 
        url_geoloc = f"https://geocoding-api.open-meteo.com/v1/search?name={location}&count=10&language=en&format=json"
        geoloc_response = requests.get(url_geoloc) 
        data_geoloc = geoloc_response.json()

        if "results" not in data_geoloc:
            print("City not found. Please try again")
            continue
        
        
        latitude  = data_geoloc["results"][0]["latitude"]
        longitude = data_geoloc["results"][0]["longitude"]
        
        coords = {
            "city": location,
            "latitude": latitude,
            "longitude": longitude
        }
        return coords
    
def get_Weather():
    
    coords = get_Coords()
    
    url_weather = f"https://api.open-meteo.com/v1/forecast?latitude={coords['latitude']}&longitude={coords['longitude']}&current=temperature_2m,relative_humidity_2m,apparent_temperature,precipitation,rain,showers,snowfall,wind_speed_10m,wind_direction_10m,wind_gusts_10m,surface_pressure,pressure_msl,cloud_cover,weather_code,is_day"
    #print(url_weather)
    weather_response = requests.get(url_weather) 
    data_weather = weather_response.json()
    #print(data_weather.keys())
    #print("Temperature:", data_weather["current"]["temperature_2m"],data_weather["current_units"]["temperature_2m"])


    weather = {
        "city": coords["city"],
        "temperature": data_weather["current"]["temperature_2m"],
        "apparent_temperature": data_weather["current"]["apparent_temperature"],
        "relative_humidity": data_weather["current"]["relative_humidity_2m"],
        "wind_speed": data_weather["current"]["wind_speed_10m"],
        "wind_direction": data_weather["current"]["wind_direction_10m"],
        "precipitation": data_weather["current"]["precipitation"],
        "surface_pressure": data_weather["current"]["surface_pressure"],
        "cloud_cover": data_weather["current"]["cloud_cover"],
    }


    weather_units = {
        "city": coords["city"],
        "temperature": data_weather["current_units"]["temperature_2m"],
        "apparent_temperature": data_weather["current_units"]["apparent_temperature"],
        "relative_humidity": data_weather["current_units"]["relative_humidity_2m"],
        "wind_speed": data_weather["current_units"]["wind_speed_10m"],
        "wind_direction": data_weather["current_units"]["wind_direction_10m"],
        "precipitation": data_weather["current_units"]["precipitation"],
        "surface_pressure": data_weather["current_units"]["surface_pressure"],
        "cloud_cover": data_weather["current_units"]["cloud_cover"],
    }
    
    return weather, weather_units

def format_Weather(weather,weather_units):

    print(f"City: {weather['city']}")
    print(f"Temperature: {weather['temperature']}{weather_units['temperature']}")
    print(f"Apparent Temperature: {weather['apparent_temperature']} {weather_units['apparent_temperature']}")
    print(f"Relative Humidity: {weather['relative_humidity']}{weather_units['relative_humidity']}")
    print(f"Wind Speed:, {weather['wind_speed']}{weather_units['wind_speed']}")
    print(f"Wind Direction:, {weather['wind_direction']}{weather_units['wind_direction']}")
    print(f"Precipitation:, {weather['precipitation']}{weather_units['precipitation']}")
    print(f"Surface Pressure:, {weather['surface_pressure']}{weather_units['surface_pressure']}")
    print(f"Cloud Cover:, {weather['cloud_cover']}{weather_units['cloud_cover']}")


def main():
    
    while True:

        weather, weather_units = get_Weather()
        format_Weather(weather, weather_units)


        again = input("Would you like to check the weather of another city? (Y/N)")

        if again.strip().lower() == "y": 
            
            continue
        else:
            break


main()
#get_Coords()
#get_Weather()
```
